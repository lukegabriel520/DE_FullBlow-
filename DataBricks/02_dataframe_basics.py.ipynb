```python
rows = [
    (1, "ONLINE",  "Apparel",      2, 50.00, "COMPLETED"),
    (2, "ONLINE",  "Electronics",  1, 40.00, "COMPLETED"),
    (3, "RETAIL",  "Electronics",  3, 36.00, "COMPLETED"),
    (4, "RETAIL",  "Home",         1, 45.00, "COMPLETED"),
    (5, "RETAIL",  "Apparel",      1, 25.00, "COMPLETED"),
    (6, "ONLINE",  "Electronics",  2, 80.00, "PENDING"),
    (7, "ONLINE",  "Home",         4, 72.00, "CANCELLED"),
    (8, "RETAIL",  "Apparel",      1, 25.00, "COMPLETED"),
    (9, "RETAIL",  "Home",         2, 36.00, "COMPLETED"),
    (10,"RETAIL",  "Electronics",  1, 40.00, "COMPLETED"),
]
cols = ["sales_key", "channel", "category", "quantity", "line_amount", "status"]
```

```python
from pyspark.sql import functions as F
```

```python
df = spark.createDataFrame(rows, cols)

filtered_df = df.filter(F.col("status") == 'COMPLETED')
final_df = (filtered_df.groupBy("channel").
            agg(F.sum("quantity").alias("units")))
```
